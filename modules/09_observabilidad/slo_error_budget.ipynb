# SLO y presupuesto de error

Calcule el presupuesto antes de diseñar alertas. Los datos de ejemplo son sintéticos y no representan producción.

```python
slo = 0.995
ventana_dias = 28
minutos_ventana = ventana_dias * 24 * 60
presupuesto_minutos = minutos_ventana * (1 - slo)
round(presupuesto_minutos, 1)
```

El SLI de ejemplo considera bueno un evento no 5xx completado en menos de 400 ms. Defina explícitamente exclusiones, fuente y ventana.

```python
periodos = ['semana 1', 'semana 2', 'semana 3', 'semana 4']
eventos_validos = [10000, 12000, 11000, 13000]
eventos_malos = [20, 35, 90, 40]
consumo_pct = [malos / validos * 100 for malos, validos in zip(eventos_malos, eventos_validos)]
list(zip(periodos, consumo_pct))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(periodos, consumo_pct, marker='o', color='#163A5F')
ax.axhline((1 - slo) * 100, color='#A61B1B', linestyle='--', label='presupuesto proporcional')
ax.set_ylabel('Eventos malos (%)')
ax.set_title('Datos sintéticos: consumo semanal')
ax.legend()
plt.show()
```
