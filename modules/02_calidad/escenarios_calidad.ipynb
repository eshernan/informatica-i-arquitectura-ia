# Escenarios de calidad y decisiones

Este notebook transforma objetivos vagos en medidas. Los datos iniciales son **ilustrativos**; reemplácelos por observaciones del experimento y documente máquina, versión, carga y repeticiones.

## Escenario

Cuando 40 usuarios concurrentes consultan disponibilidad durante operación normal, LibreReserva responde al menos el 95 % de solicitudes en menos de 400 ms y mantiene menos de 1 % de errores durante 10 minutos.

```python
alternativas = {
    'monolito_modular': {'p95_ms': 280, 'errores_pct': 0.2, 'complejidad': 2},
    'servicios_separados': {'p95_ms': 360, 'errores_pct': 0.6, 'complejidad': 5},
}
umbral_ms = 400
for nombre, medida in alternativas.items():
    medida['cumple'] = medida['p95_ms'] <= umbral_ms and medida['errores_pct'] < 1
alternativas
```

```python
import matplotlib.pyplot as plt

nombres = list(alternativas)
latencias = [alternativas[n]['p95_ms'] for n in nombres]
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(nombres, latencias, color=['#18794E', '#B45309'])
ax.axhline(umbral_ms, color='#A61B1B', linestyle='--', label='umbral')
ax.set_ylabel('Latencia p95 (ms)')
ax.set_title('Datos ilustrativos: reemplazar con mediciones')
ax.legend()
plt.show()
```

## Interpretación

Ambas alternativas cumplen el umbral de ejemplo. La medición no demuestra que distribuir sea conveniente: la decisión también considera modificabilidad, operación, costo y capacidad del equipo. Escriba una conclusión limitada y una amenaza a la validez.
