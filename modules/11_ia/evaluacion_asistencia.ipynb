# Evaluación controlada de asistencia de IA

El objetivo no es declarar un ganador universal. Se comparan condiciones bajo una tarea, entorno y conjunto de pruebas fijos. No incluya datos personales, secretos ni código restringido.

```python
import pandas as pd

datos = pd.DataFrame([
    {'tarea': 'intervalo', 'condicion': 'humana', 'minutos': 18, 'pruebas_ok': 8, 'defectos': 0},
    {'tarea': 'intervalo', 'condicion': 'asistida', 'minutos': 12, 'pruebas_ok': 7, 'defectos': 1},
    {'tarea': 'contrato', 'condicion': 'humana', 'minutos': 25, 'pruebas_ok': 10, 'defectos': 0},
    {'tarea': 'contrato', 'condicion': 'asistida', 'minutos': 17, 'pruebas_ok': 10, 'defectos': 0},
])
datos
```

Los datos anteriores son sintéticos. Sustitúyalos por cinco tareas como mínimo. El tiempo incluye solicitud, revisión, corrección y validación; medir solo generación favorece artificialmente a la asistencia.

```python
resumen = datos.groupby('condicion').agg(
    minutos_promedio=('minutos', 'mean'),
    pruebas_aprobadas=('pruebas_ok', 'sum'),
    defectos=('defectos', 'sum'),
)
resumen
```

```python
ax = datos.pivot(index='tarea', columns='condicion', values='minutos').plot.bar(
    color=['#163A5F', '#18794E'], figsize=(7, 4), rot=0
)
ax.set_ylabel('Tiempo total (minutos)')
ax.set_title('Ejemplo sintético; no generalizar')
```
